**Tasa de conversión**

Se transformó la variable objetivo en formato binario (0 = no conversión, 1 = conversión) para facilitar el análisis estadístico y modelado posterior.

```python
df_final["y_bin"] = df_final["y"].map({"yes": 1, "no": 0})


conversion_rate = df_final["y_bin"].mean()
print(f"Tasa global de conversión: {conversion_rate:.2%}")
```

3.3 Contacto previo

He visto que si no agrego bien la variable "pdays", puede inducir a error en la tasa de conversión, por lo que he tenido que utilizar valores numéricos originales para definir los grupos de contacto previo.

```python
df_final["pdays_clean"] = df_final["pdays"].replace(999, np.nan)
```

```python
df_final["pdays_group"] = pd.cut(
    df_final["pdays_clean"],
    bins=[0,7,30,90,365,9999],
    labels=["≤7d","8-30d","31-90d","91-365d",">365d"]
)

df_final["pdays_group"] = df_final["pdays_group"].cat.add_categories("sin contacto")
df_final.loc[df_final["pdays"].isna(), "pdays_group"] = "sin contacto"
```

```python
order = ["sin contacto","≤7d","8-30d","31-90d","91-365d",">365d"]

plt.figure(figsize=(10,5))
sns.barplot(data=df_final, x="pdays_group", y="y_bin", order=order, estimator="mean")
plt.title("Conversión según último contacto previo")
plt.xlabel("Último contacto previo")
plt.ylabel("Tasa de conversión")
plt.show()
```

Los dos gráficos que hay a continuación no se tienenen cuenta para el informe, pero los dejo para comprobar como si agrupo mal la etiqueta "pdays", puede inducir al error con la tasa de conversión.

```python
plt.figure(figsize=(12,6))
sns.barplot(data=df_final, x="pdays_label", y="y_bin", estimator="mean")
plt.title("Conversión según contacto previo")
plt.show()
```

```python
df_final["pdays_num_order"] = (
    df_final["pdays_label"]
    .replace("sin contacto previo", -1)
    .astype(int)
)
```

```python
df_final["pdays_group"] = pd.cut(
    df_final["pdays_num_order"],
    bins=[-1,0,7,30,90,365,999],
    labels=["sin contacto","≤7d","8-30d","31-90d","91-365d",">365d"]
)

plt.figure(figsize=(10,5))
sns.barplot(data=df_final, x="pdays_group", y="y_bin", estimator="mean")
plt.title("Conversión según antigüedad del último contacto")
plt.xlabel("Último contacto previo")
plt.ylabel("Tasa de conversión")
plt.show()
```
