```python
# Tratamiento de datos

import pandas as pd
pd.set_option('display.max_columns', None)

import numpy as np

# ----------

# Exploración de archivos
import sys
sys.path.append('../../../')

#----------

# Funciones personalizadas
from src import sp_cleaning as sc
```

Cargamos los datos y hacemos una copia

```python
df_bank = pd.read_csv('../../../data/bank-additional.csv')
```

```python
df_clean = df_bank.copy()
```

Elimino la columna "Unnamed" porque no aporta información ni valor al dataframe

```python
sc.drop_columns(df_clean, ["Unnamed: 0"])
```

Convierto algunas columnas dtype object a columnas dytpe numéricas

```python
sc.comas(df_clean)
```

Cambio el formato de la columna "edad" a "int64" porque la edad se mide en número enteros, no en decimales.

```python
df_clean['age'] = df_clean['age'].astype("Int64")
```

Cambio a minúscula todas las columnas categóricas del dataframe

```python
sc.minus(df_clean)
```

Normalizamos el texto de las columnas del dataset para una mejor comprensión

```python
sc.normalizar_texto(df_clean)
```

Cambiamos el nombre de algunas columnas 

```python
df_clean.columns = df_clean.columns.str.replace('.', '_').str.lower()
df_clean = df_clean.rename(columns={'id_': 'id'})
df_clean = df_clean.rename(columns={'euribor3m': 'euribor_3m'})
```

La columna 'date' es del tipo 'object', por lo que la cambiamos a tipo 'datetime'. Como los meses están en español, los pasamos a ingles porque es el lenguaje que reconoce python. Primero sustituyo "_" por "-" y luego creamos un diccionario y luego lo reemplazamos.

```python
df_clean['date'] = df_clean['date'].str.replace('_', '-')
```

```python
sc.convierte_datetime(df_clean, 'date')

```

Convertimos a tipo booleanos las columnas "default", "housing", "loan".

```python
sc.boolean(df_clean,'default')
sc.boolean(df_clean,'housing')
sc.boolean(df_clean,'loan')
```

Ahora, creamos nuevas columnas para convertir los valores booleanos en una columnas "object" para su análisis.

```python
sc.boolean_to_str(df_clean, 'default', 'default_str')
sc.boolean_to_str(df_clean, 'housing', 'housing_str')
sc.boolean_to_str(df_clean, 'loan', 'loan_str')
```

Extraemos los datos de contacto (mes y año) de la columna "date". Para ellos generamos dos nuevas columnas que sean "contact_month" y "contact_year"

```python
df_clean['contact_month'] = df_clean['date'].dt.month_name().str.lower()
df_clean['contact_year'] = df_clean['date'].dt.year.astype('Int64')
```

En la columna "contact", aparece el valor "999", que entiendo que significa "Sin contacto". Voy a crear una nueva columna para sustituir ese valor por un valor "nulo" para que no desvirtue los datos de la media o mediana en en eda numérico.

```python
df_clean['pdays_clean'] = df_clean['pdays'].replace(999, np.nan).astype('Int64')
media_pdays = df_clean['pdays_clean'].mean()
```

Crear y guardar el nuevo archivo limpio 

```python
df_clean.to_csv('../../../data/data_bank_clean.csv', index=False)

```
