Como ya tenemos los datos del banco limpios y el dataset de customer_details no tiene nulos, lo que vamos a hacer ahora es unir ambos dataset para así poder realizar el análisis final.

```python
df_bank = pd.read_csv('../../../data/data_bank_final.csv')
df_customer = pd.read_csv('../../../data/data_customer_clean.csv')
```
