```python
sc.normalizar_texto(df_clean)
```

Cambiamos el nombre de algunas columnas más específicas que no están dentro de la función "normalizar_texto"

```python
df_clean.columns = df_clean.columns.str.replace('.', '_').str.lower()
df_clean = df_clean.rename(columns={'id_': 'id'})
df_clean = df_clean.rename(columns={'euribor3m': 'euribor_3m'})
```

```python
df_clean['date'] = df_clean['date'].str.replace('_', '-')
```
