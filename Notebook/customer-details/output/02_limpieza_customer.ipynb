```python
# Tratamiento de datos

import pandas as pd
pd.set_option('display.max_columns', None)

import numpy as np

# ----------

# Exploración de archivos
import sys
sys.path.append('../../../')

#----------

# Funciones personalizadas
from src import sp_cleaning as sc
```

Cargamos los datos y hacemos una copia

```python
xls = pd.ExcelFile('../../../data/customer-details.xlsx')

df_customer = pd.concat(
    [xls.parse(sheet) for sheet in xls.sheet_names],
    ignore_index=True
)

df_clean = df_customer.copy()
```

Elimino la columna "Unnamed" porque no aporta información ni valor al dataframe

```python
sc.drop_columns(df_clean, ["Unnamed: 0"])
```

Cambio a minúscula el nombre de todas las columnas

```python
df_clean.columns = df_clean.columns.str.lower()
```

Normalizo el texto "numwebvisitsmonth" y lo cambio por "num_web_visits_month" para que quede más claro.

```python
df_clean = df_clean.rename(columns={'numwebvisitsmonth': 'num_web_visits_month'})
```

Crear dos columnas con el mes y el año que el cliente se convirtió en cliente de la empresa

```python
df_clean['contact_month'] = df_clean['dt_customer'].dt.month_name().str.lower()
df_clean['contact_year'] = df_clean['dt_customer'].dt.year.astype('Int64')
```

Crear y guardar el nuevo archivo limpio 

```python
df_clean.to_csv('../../../data/data_customer_clean.csv', index=False)
```
