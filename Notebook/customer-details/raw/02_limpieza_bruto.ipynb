Cargamos los datos y hacemos una copia

```python
xls = pd.ExcelFile('../../../data/customer-details.xlsx')

df_customer = pd.concat(
    [xls.parse(sheet) for sheet in xls.sheet_names],
    ignore_index=True
)
```

```python
df_clean = df_customer.copy()
```
